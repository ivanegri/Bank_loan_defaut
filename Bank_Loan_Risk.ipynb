## Preprocessing:

### How to handle missing values in credit_score?

```python
df.loc[df['default']==1,'credit_score'] = df['credit_score']*0.30
```

* As I created this Data set,i decided to low credit_score with defaul to 30% from what it was

```python
df.previous_defaults.value_counts()
```

```python
df.loc[df['previous_defaults']==1,'credit_score'] = df['credit_score']*0.80
```

```python
df.loc[df['previous_defaults']==2,'credit_score'] = df['credit_score']*0.50
```

* Did almost the same with previous default,cause it impact on the credit_score (using mu knowledge).
* PS: I'm only doing this cause I created this Data Base
* Now, I can fill the missing values of the credit Score, based on the mean times percents I used with default

```python
df.isna().sum()
```

```python
df.loc[(df['default'] == 1) & (df['credit_score'].isna()), 'credit_score'] = df['credit_score'].mean() * 0.3
```

```python
df.loc[(df['previous_defaults'] == 1) & (df['credit_score'].isna()), 'credit_score'] = df['credit_score'].mean() * 0.8
```

```python
df.loc[(df['previous_defaults'] == 2) & (df['credit_score'].isna()), 'credit_score'] = df['credit_score'].mean() * 0.5
```

```python
df.loc[ (df['credit_score'].isna()), 'credit_score'] = df['credit_score'].mean()
```

```python
df.isna().sum()
```

* OK, treated the credit_score, trying to keep it close to real world

* The loan amount requested should be at least equal or higher than loan amount

```python
df.loc[(df['default']) ==0 & (df['previous_defaults']==0),'loan_amount_requested']=df['loan_amount']
```

```python
df.loc[(df['loan_amount_requested']) < (df['loan_amount']),'loan_amount'].count()
```

```python
df.loc[(df['default']) ==1 & (df['previous_defaults']>0),'loan_amount_requested']=df['loan_amount']*1.5
```

```python
df.loc[(df['loan_amount_requested']) < (df['loan_amount']),'loan_amount'].count()
```

```python
df.loc[(df['default']) ==1 & (df['previous_defaults']<=2),'loan_amount_requested']=df['loan_amount']*1.4
```

```python
df.loc[(df['loan_amount_requested']) < (df['loan_amount']),('loan_amount','loan_amount_requested','previous_defaults','default')]
```

* the monthly income, should be higher than the installment, and the installment should be loan amount,times interests, divided by loan term
* Let's adjust installment

```python
df['installment'] = df.apply(
    lambda row: row['loan_amount'] * 
    (row['interest_rate'] * (1 + row['interest_rate'])**row['loan_term']) / 
    ((1 + row['interest_rate'])**row['loan_term'] - 1),
    axis=1
)
```

```python
df['installment'] = round(df['installment'],2)
```

```python
df.loc[((df['mothly_income']+df['investment_income'])) > (df['installment']),'mothly_income'] = df['mothly_income']
```

```python
df.loc[((df['mothly_income']+df['investment_income'])) < (df['installment']),'mothly_income'] = df['mothly_income']*6.5
```

```python
df.loc[((df['mothly_income'])) < (df['installment']),'default'] = 1
```

```python
df.loc[((df['mothly_income']+df['investment_income'])) > (df['installment']),'default'] = 0
```


* Looks like it tested well


```python
df['default'].value_counts(normalize=True).mul(100)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("\nShape dos dados:")
print(f"X_train: {X_train.shape} | y_train: {y_train.shape}")
print(f"X_test:  {X_test.shape}  | y_test:  {y_test.shape}")

print("\nExemplo de X_train:")
print(X_train.head(3))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

# Treinar modelo
model = XGBClassifier(
    scale_pos_weight=(len(y_train.default) - sum(y_train.default)) / sum(y_train.default),  # Balanceamento
    eval_metric='aucpr',
    random_state=42
)
model.fit(X_train_scaled, y_train)

# Avaliar
y_pred = model.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
```
